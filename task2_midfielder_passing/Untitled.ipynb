```python
df["Group"].value_counts()
```

```python
starters = df[df["Group"] == "Starter"].sample(n=50, random_state=42)

substitutes = df[df["Group"] == "Substitute"].sample(n=50, random_state=42)

sample100 = pd.concat([starters, substitutes])
```

```python
sample100["Group"].value_counts()
```

```python
sample100.groupby("Group")["Min"].describe()
```

```python
import scipy.stats as stats
import numpy as np

starter_min = sample100[sample100["Group"] == "Starter"]["Min"]
sub_min = sample100[sample100["Group"] == "Substitute"]["Min"]

starter_ci = stats.t.interval(
    0.95,
    len(starter_min)-1,
    loc=np.mean(starter_min),
    scale=stats.sem(starter_min)
)

sub_ci = stats.t.interval(
    0.95,
    len(sub_min)-1,
    loc=np.mean(sub_min),
    scale=stats.sem(sub_min)
)

print("Starter 95% CI:", starter_ci)
print("Substitute 95% CI:", sub_ci)
```

```python
t_stat, p_value = stats.ttest_ind(starter_min, sub_min, equal_var=False)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```
